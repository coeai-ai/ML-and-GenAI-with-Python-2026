```python
name = input("Enter student name: ")
roll = input("Enter roll number: ")

n = int(input("Enter number of subjects: "))

total = 0

for i in range(n):
    marks = float(input("Enter marks: "))
    total += marks

percentage = total / n

if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "F"

print("\nStudent Name:", name)
print("Roll Number:", roll)
print("Percentage:", percentage)
print("Grade:", grade)
```
