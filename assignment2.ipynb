```python
n = int(input("Enter number of students: "))

for i in range(n):
    print("\nStudent", i + 1)

    name = input("Enter student name: ")
    roll = input("Enter roll number: ")

    total = 0

    for j in range(5):
        marks = float(input(f"Enter marks of subject {j + 1}: "))
        total += marks

    percentage = total / 5

    if percentage >= 90:
        grade = "A+"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    print("\nStudent Details")
    print("Name:", name)
    print("Roll No:", roll)
    print("Percentage:", percentage)
    print("Grade:", grade)
```
