```python
l=float(input("Enter length: "))
b=float(input("Enter breadth: "))
area_rect=l*b
print("Area of rectangle =", area_rect)
```

```python
principal=float(input("Enter principal amount (in Rs.): "))
rate=float(input("Enter rate of interest per annum (in %): "))
time=float(input("Enter time period (in yrs): "))
simple_interest=principal*rate*time
print("Simple interest = Rs.", simple_interest)
```

```python
celsius=float(input("Enter temperature (in degree Celsius): "))
fahrenheit=(celsius*9/5)+32
print("Temperature =", fahrenheit, "degree Fahrenheit")
```

```python
n=int(input("Enter number: "))
print("Square of", n, "=", n**2)
print("Cube of", n, "=", n**3)
```

```python
#Enter details
name=input("Enter student name: ")
roll_number=int(input("Enter roll number: "))
#Enter marks in different variables
math, sci, eng=eval(input("Enter marks for Maths, Science, English: "))
#Marks calculate
total_marks=math+sci+eng
avg_marks=total_marks/300
percentage=avg_marks*100
#Report card
print()
print("STUDENT REPORT CARD")
print("================================")
print("Name       :", name)
print("Roll No.   :", roll_number)
print("--------------------------------")
print("Maths      :", math)
print("Science    :", sci)
print("English    :", eng)
print("--------------------------------")
print("Total Marks:", total_marks)
print("Percentage :", percentage)
```
