```python
#create a function to calculate sum of first n natural numbers
def sum_nat_nums(n):
    sum=0
    for i in range(1,n+1):
        sum+=i
    return sum
n=int(input("enter the value of n:"))
sum_nat_nums(n)
```

```python
#create a function to reverse a number
def rev_num(n):
    rev=0
    while n>0:
        rem=n%10
        rev=rev*10+rem
        n=n//10
    return rev
n=int(input("enter the number:"))
print("original:",n)
print("reversed:",rev_num(n))
```

```python
#create a function to count digits in a number
def count_digits(n):
  c=0
  while n>0:
    c+=1
    n=n//10
  return c
n=int(input("enter the number"))
print("number of digits:",count_digits(n))
```

```python
#create a function to check if a number is a palindrome
def palindrome(n):
  rev=0
  temp=n
  while temp>0:
    rem=temp%10
    rev=rev*10+rem
    temp=temp//10
  if n==rev:
    print(n,"is a palindrome number")
  else:
    print(n,"is not a palindrome number")
n=int(input("enter the number"))
palindrome(n)
```

```python
#create a function to generate fibonacci series
def fibonacci(n):
  a,b=0,1
  print(a,b,end=" ")
  for i in range(1,n-2+1):
    a,b=b,a+b
    print(b,end=" ")
n=int(input("enter number of terms"))
fibonacci(n)
```

```python
#create a calculator
def add(a,b):
  return a+b
def subtract(a,b):
  return a-b
def multiply(a,b):
  return a*b
def divide(a,b):
  if b==0:
    print("error! division by zero")
  return a/b
print("SIMPLE CALCULATOR")
a=float(input("enter first number"))
b=float(input("enter second number"))
print("select operation \n1. add\n2. subtract\n3. multiply\n4. divide")
choice=int(input("enter your choice (1-4):"))
if choice==1:
  print("result:",add(a,b))
elif choice==2:
  print("result:",subtract(a,b))
elif choice==3:
  print("result:",multiply(a,b))
elif choice==4:
  print("result:",divide(a,b))
else:
  print("invalid input")
```

```python
#handle division error
def safe_division():
    try:
        numerator = float(input("enter numerator:"))
        denominator = float(input("enter denominator:"))
        result = numerator / denominator
        print("result of division:",result)
    except ZeroDivisionError:
        print("exception caught: you cannot divide a number by zero!")
    except ValueError:
        print("exception caught: please enter valid numbers")
safe_division()
```

```python
#create a student class with name and marks
class Student:
    #constructor to initialize name and marks
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks  # Expecting a list of marks
    #method to display student details
    def display_details(self):
        print("--- student details ---")
        print("student name:",self.name)
        print("marks obtained:",self.marks)
        print("total marks:",sum(self.marks))
#creating an object of the Student class
student1=Student("anjali singh", [85, 92, 78])
#displaying details using the object method
student1.display_details()
```
