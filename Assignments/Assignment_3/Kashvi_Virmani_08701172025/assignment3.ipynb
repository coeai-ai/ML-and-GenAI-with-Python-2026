## Q1. Create a function to print first 10 natural numbers

```python
def print_natural_numbers():
    for i in range(1,11):
        print(i)
print_natural_numbers()
```

## Q2. Write a function to calculate the sum of first 10 natural numbers.

```python
def sum_of_natural_numbers():
    sum = 0
    for i in range(1,11):
        sum += i
    print(f"Sum of first 10 natural numbers is {sum}")
sum_of_natural_numbers()
```

## Q3. Create a function to reverse a number

```python
def reverse_number(num):
    n = num
    reverse_num = 0
    while(n>0):
        reverse_num = reverse_num*10 + (n%10)
        n = n // 10
    return reverse_num

print(reverse_number(28461))
```

## Q4. Count digits of a number

```python
def count_digits(num):
    n = num
    cnt = 0
    while(n>0):
        cnt += 1
        n = n // 10
    return cnt
print(count_digits(124835))
```

## Q5. Write function to check palindrome number

```python
def check_palindrome(num):
    n = num
    reverse_num = 0
    while(n>0):
        reverse_num = reverse_num*10 + (n%10)
        n = n // 10
    if(reverse_num == num):
        return True
    else:
        return False
print(check_palindrome(112))
```

## Q6. Write a function to generate fibonacci series

```python
def fibonacci_series(no_of_terms):
    x = 0
    y = 1
    z = 0
    while(no_of_terms>0):
        print(z,end=" ")
        x = y
        y = z
        z = x + y
        no_of_terms -= 1

no_of_terms = int(input("Enter number of terms you want :- "))
fibonacci_series(no_of_terms)
```

## Q7. Calculator using functions that contains the following features. -User selects operations -Program performs calculations -Display result

```python
def addition(a,b):
    return (a+b)
def subtraction(a,b):
    return (a-b)
def multiplication(a,b):
    return (a*b)
def division(a,b):
    return (a/b)
print("---------- Calculator ----------")
n1 = float(input("Enter first number :- "))
n2 = float(input("Enter second number :- "))
op = input("Enter operation to perform\n+ -> Addition\n- -> Subtraction\n* -> Multiplication\n/ -> Division\n")
if(op=="+"):
    print(f"Answer is {addition(n1,n2)}")
elif(op=="-"):
    print(f"Answer is {subtraction(n1,n2)}")
elif(op=="*"):
    print(f"Answer is {multiplication(n1,n2)}")
elif(op=="/"):
    print(f"Answer is {division(n1,n2)}")
```

## Q8. Create a text file and store student details

```python
with open("student.txt", "w") as f:
    header = "Roll_No" + "\t" + "Age" + "\t" + "Marks" + "\n"
    f.write(header)
    while(True):
        r = int(input("Enter roll no :- "))
        a = int(input("Enter age :- "))
        m = int(input("Enter marks :- "))
        data = str(r) + "\t" + str(a) + "\t" + str(m) + "\n"
        f.write(data)

        ch = input("Do you want to add more (y/n) :- ")
        if(ch=="n"):
            break
    print("Data succesfully inserted...")
```

## Q10. Handling division by zero using exception handling

```python
n1 = int(input("Enter first number : "))
n2 = int(input("Enter second number : "))
try:
    print(n1/n2)
except(ZeroDivisionError):
    print("Error : Division by zero not possible")
else:
    print("No Error")
finally:
    print("Program terminated succesfully")
```

## Q11. Create a student class with name and marks

```python
class Student:
    def __init__(self,name,marks): # constructor
        self.name = name
        self.marks = marks

s1 = Student("Kashvi",90)
s2 = Student("Jai",89)
print(s1.name)
print(s1.marks)
print(s2.name)
print(s2.marks)
```
