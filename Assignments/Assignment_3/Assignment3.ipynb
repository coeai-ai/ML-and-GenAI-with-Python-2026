```python
def sumN(n):
  sum=0
  for i in range(1, n+1):
    sum+=i
  return sum
num=int(input("Enter number: "))
sum=sumN(num)
print(sum)
```

```python
def rev(n):
  rev=0
  while n>0:
    digit=n%10
    rev=rev*10+digit
    n=n//10
  return rev
num=int(input("Enter number: "))
reverse_num=rev(num)
print(reverse_num)
```

```python
def digitsN(n):
  count=0
  while n>0:
    n=n//10
    count+=1
  return count
num=int(input("Enter number: "))
n_digits=digitsN(num)
print(n_digits)
```

```python
def palindrome(n):
  temp=n
  rev=0
  while n>0:
    digit=n%10
    rev=rev*10+digit
    n=n//10
  if rev==temp:
    return True
  else:
    return False
num=int(input("Enter number: "))
is_palindrome=palindrome(num)
print(is_palindrome)
```

```python
def fibonacciN(n):
  if n==0:
    print(0)
    return
  a, b=0, 1
  print(a, b, end=" ")
  for i in range(0, n-2):
    a, b=b, a+b
    print(b, end=" ")
  print()
num=int(input("Enter number: "))
fibonacciN(num)
```

```python
def add(x, y):
  return x+y
def subtract(x, y):
  return x-y
def multiply(x, y):
  return x*y
def divide(x, y):
  if y!=0:
    return x/y
  else:
    return "Division by zero error"
def calculator():
    print("1. Add (+)\n2. Subtract (-)\n3. Multiply (*)\n4. Divide (/)")
    choice=input("Enter choice 1, 2, 3, or 4: ")
    if choice in ["1", "2", "3", "4"]:
        try:
            num1=float(input("Enter first number: "))
            num2=float(input("Enter second number: "))
        except ValueError:
            print("Invalid input")
            return
        if choice=="1":
            result=add(num1, num2)
            op="+"
        elif choice=="2":
            result=subtract(num1, num2)
            op="-"
        elif choice=="3":
            result=multiply(num1, num2)
            op="*"
        elif choice=="4":
            result=divide(num1, num2)
            op="/"
        print(num1, op, num2, "=", result)
calculator()
```
