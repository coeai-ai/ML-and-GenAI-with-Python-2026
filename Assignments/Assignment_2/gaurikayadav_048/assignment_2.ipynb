```python
#find sum of first 10 natural numbers
sum=0
for i in range(1,11):
  sum+=i
print("sum of first 10 natural numbers is:",sum)
```

```python
#find factorial of a number
n=int(input("enter a number:"))
fact=1
for i in range(1,n+1):
  fact*=i
print("factorial of",n,"is:",fact)
```

```python
#print fibonacci series
n=int(input("enter the number of terms:"))
a,b=0,1
print(a,b,end=" ")
for i in range(1,n-2+1):
  a,b=b,a+b
  print(b,end=" ")
```

```python
#find largest among three numbers
n1=int(input("enter first number:"))
n2=int(input("enter second number:"))
n3=int(input("enter third number:"))
largest = max(n1, n2, n3)
print("largest number is:",largest)
```

```python
#create student result system
n=int(input("enter the number of students:"))
for i in range(1,n+1):
  print("enter details of student",i)
  name=input("enter name:")
  rollno=input("enter roll no:")
  s1=float(input("enter marks in subject 1:"))
  s2=float(input("enter marks in subject 2:"))
  s3=float(input("enter marks in subject 3:"))
  t=s1+s2+s3
  p=t/3
  if p<=100 and p>90:
    g="A+"
  elif p<=90 and p>80:
    g="A"
  elif p<=80 and p>70:
    g="B+"
  elif p<=70 and p>60:
    g="B"
  elif p<=60 and p>50:
    g="C"
  elif p<=50 and p>40:
    g="D"
  elif p<=40 and p>30:
    g="E"
  else:
    g="F"
  print("STUDENT REPORT")
  print("name:",name)
  print("roll no:",rollno)
  print("total:",t," / 300.0")
  print("percentage:",p)
  print("grade:",g)
```
