```python
#find area of rectangle
l=float(input("enter length of rectangle in cm:"))
b=float(input("enter breadth of rectangle in cm:"))
area=l*b
print("area of rectangle (sq. cm) is:",area)
```

```python
#find simple interest
p=float(input("enter principal amount:"))
r=float(input("enter rate of interest:"))
t=float(input("enter time in years:"))
si=(p*r*t)/100
print("simple interest is:",si)
```

```python
#convert temp from celsius to farenheit
c=float(input("enter temperature in celsius:"))
f=(c*9/5)+32
print("temperature in farenheit is:",f)
```

```python
#calculate avg of three numbers
n1=int(input("enter first number:"))
n2=int(input("enter second number:"))
n3=int(input("enter third number:"))
avg=(n1+n2+n3)/3
print("average of three numbers is:",avg)
```

```python
#find square and cube of a number
n=int(input("enter a number:"))
sq=n*n
cube=n*n*n
print("square of the number is:",sq)
print("cube of the number is:",cube)
```

```python
#swap two numbers without a third variable
n1=int(input("enter first number:"))
n2=int(input("enter second number:"))
print("before swapping:")
print("first number:",n1)
print("second number:",n2)
n1,n2=n2,n1
print("after swapping:")
print("first number:",n1)
print("second number:",n2)
```

```python
#create a student report program
name=input("enter student name:")
sub1=float(input("enter marks in subject 1:"))
sub2=float(input("enter marks in subject 2:"))
sub3=float(input("enter marks in subject 3:"))
sub4=float(input("enter marks in subject 4:"))
sub5=float(input("enter marks in subject 5:"))
total=sub1+sub2+sub3+sub4+sub5
avg=total/5
print("student report:")
print("name:",name)
print("total marks:",total," / 500.0")
print("average marks:",avg)
```
