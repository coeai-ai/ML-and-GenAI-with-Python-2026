```python
#Ques1)Find the area of rectangle
length = int(input("Length of rectangle :"))
width = int(input("Width of rectangle :"))
area = length*width
print(f"Area of rectangle is :{area}")
```

```python
#Ques2)Find Simple interest
principal_val = int(input("Enter principal value:"))
rate = int(input("Enter the rate:"))
time = int(input("Enter time:"))
simple_interest = (principal_val*rate*time)/100
print(f"The simple interest is:{simple_interest}")
```

```python
#Ques3)Convert temperature from celsius to fahrenhit
temp_celsius = int(input("Enter the temperature in celsius:"))
temp_fahrenhit = (temp_celsius*9/5)+32
print(f"Temperature in fahrenhit is:{temp_fahrenhit}")
```

```python
#Ques4)Calculate average of three numbers
num_1=float(input("Enter your first number:"))
num_2=float(input("Enter your second number:"))
num_3=float(input("Enter your third number:"))
average = (num_1+num_2+num_3)/3
print(f"The average of given three numbers is:{average}")
```

```python
#Ques5)Find square and cube of the given numbers
num = float(input("Enter your number:"))
num_squared = num**2
num_cube = num**3
print(f"The square of given number is:{num_squared} \nThe cube of the given number is:{num_cube}")
```

```python
#Ques6)Swap two numbers without using third variable
x = float(input("Enter number x:"))
y = float(input("Enter number y:"))
x, y = y, x
print(f"x ;{x} \ny:{y}")
```

```python
#Ques7)Student Report Program
#Taking student details
print("STUDENT REPORT")
student_name = input("Enter your name:")
roll_name = input("Enter roll no:")
#Storing marks as variables
maths_marks = float(input("Enter your marks in mathematics:"))
science_marks = float(input("Enter your marks in science:"))
english_marks = float(input("Enter your marks in english:"))
hindi_marks = float(input("Enter your marks in hindi:"))
sst_marks = float(input("Enter your marks in sst:"))
#Calculating Total Marks scored and percentage obtained
max_marks = 500
marks_obtained = (maths_marks+science_marks+hindi_marks+english_marks+sst_marks)
total_percentage = (marks_obtained/max_marks)*100
#Displaying the whole report
print(f"Student:{student_name}")
print(f"Roll No:{roll_name}")
print(f"Marks in Maths:{maths_marks}")
print(f"Marks in Science:{science_marks}")
print(f"Marks in English:{english_marks}")
print(f"Marks in Social Science:{sst_marks}")
print(f"Marks in Hindi:{hindi_marks}")
print(f"TOTAL MARKS OBTAINED:{marks_obtained}")
print(f"TOTAL PERCENTAGE OBTAINED:{total_percentage}")
```
