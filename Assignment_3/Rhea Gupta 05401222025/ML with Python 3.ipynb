```python
# Create a function to print first 10 natural numbers.
def print_first_10_natural_numbers():
    print("First 10 natural numbers:")
    for i in range(1, 11):
        print(i, end=" ")
    print()  # New line
```

```python
# Create a function to calculate sum of first N natural numbers.
def calculate_sum_of_n_natural_numbers(n):
    total_sum = 0
    for i in range(1, n + 1):
        total_sum += i
    return total_sum
```

```python
# Create a function to reverse a number.
def reverse_number(num):
    reversed_num = 0
    temp = abs(num)  # Handle positive version for extraction

    while temp > 0:
        remainder = temp % 10
        reversed_num = (reversed_num * 10) + remainder
        temp //= 10

    # Restore negative sign if the original number was negative
    return -reversed_num if num < 0 else reversed_num
```

```python
# Create a function to check palindrome number.
def is_palindrome(num):
    # Negative numbers are generally not considered palindromes due to the '-' sign
    if num < 0:
        return False
    return num == reverse_number(num)
```

```python
# Create a function to generate Fibonacci series up to N terms.
def generate_fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]

    series = [0, 1]
    while len(series) < n:
        next_term = series[-1] + series[-2]
        series.append(next_term)
    return series
```

```python
# Calculator Using Functions
def add(a, b): return a + b
def subtract(a, b): return a - b
def multiply(a, b): return a * b
def divide(a, b): return a / b if b != 0 else "Error! Division by zero."

def calculator():
    print("\n--- Calculator Menu ---")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")

    choice = input("User selects operation (1/2/3/4): ")

    if choice in ['1', '2', '3', '4']:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))

        # Program performs calculation
        if choice == '1':
            result = add(num1, num2)
        elif choice == '2':
            result = subtract(num1, num2)
        elif choice == '3':
            result = multiply(num1, num2)
        elif choice == '4':
            result = divide(num1, num2)

        # Display result
        print(f"Result: {result}")
    else:
        print("Invalid Choice")
```

```python
import os

# 1. Create a text file and store student details
def create_and_store_student_details(filename="students.txt"):
    data = """Roll No: 1, Name: Rahul, Marks: 85
Roll No: 2, Name: Priya, Marks: 92
Roll No: 3, Name: Amit, Marks: 78
"""
    with open(filename, "w") as file:
        file.write(data)
    print(f"Data successfully saved to {filename}")


# 2. Read data from a file
def read_file_data(filename="students.txt"):
    print(f"\n--- Reading from {filename} ---")
    if not os.path.exists(filename):
        print("File does not exist.")
        return

    with open(filename, "r") as file:
        content = file.read()
        print(content)


# 3. Handle division by zero using exception handling
def safe_division():
    print("\n--- Exception Handling Division ---")
    try:
        numerator = float(input("Enter numerator: "))
        denominator = float(input("Enter denominator: "))
        result = numerator / denominator
        print(f"Result: {result}")
    except ZeroDivisionError:
        print("Error: You cannot divide a number by zero.")
    except ValueError:
        print("Error: Please enter valid numerical values.")


# 4. Create a Student class with name and marks
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display_details(self):
        print(f"Student Name: {self.name} | Marks Obtained: {self.marks}")


# Execution Demonstrations:
create_and_store_student_details()
read_file_data()

# Instantiate Student Objects
student1 = Student("Ananya", 95)
student2 = Student("Harshit", 88)
print("\n--- Student Class Output ---")
student1.display_details()
student2.display_details()
```
