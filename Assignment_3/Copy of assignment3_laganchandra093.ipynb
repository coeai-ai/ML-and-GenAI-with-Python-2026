```python
# 1) Print first 10 natural numbers
def print_first_10_natural():
    for i in range(1, 11):
        print(i)
print("1) First 10 natural numbers")
print_first_10_natural()

# 2) Sum of first n natural numbers
def sum_first_n():
    n = int(input("\n2) Enter value of n for sum: "))
    print("Sum:", n * (n + 1) // 2)
sum_first_n()

# 3) Reverse a number
def reverse_number():
    num = int(input("\n3) Enter a number to reverse: "))
    rev = 0
    temp = num
    while temp > 0:
        rev = rev * 10 + temp % 10
        temp //= 10
    print("Reversed number:", rev)

reverse_number()

# 4) Count digits in a number
def count_digits():
    num = int(input("\n4) Enter a number to count digits: "))
    if num == 0:
        print("Digits: 1")
        return
    count = 0
    temp = num
    while temp > 0:
        count += 1
        temp //= 10
    print("Digits:", count)
count_digits()

# 5) Check palindrome number
def is_palindrome():
    num = int(input("\n5) Enter a number to check palindrome: "))
    temp = num
    rev = 0
    while temp > 0:
        rev = rev * 10 + temp % 10
        temp //= 10

    if rev == num:
        print("Palindrome number")
    else:
        print("Not a palindrome number")

is_palindrome()

# 7) Fibonacci series
def fibonacci_series():
    n = int(input("\n7) Enter number of Fibonacci terms: "))
    a, b = 0, 1
    print("Fibonacci series:")
    for _ in range(n):
        print(a, end=" ")
        a, b = b, a + b
    print()

fibonacci_series()

# 8) Calculator using functions
def calculator():
    print("\n8) Calculator")
    print("1 Add")
    print("2 Subtract")
    print("3 Multiply")
    print("4 Divide")
    choice = int(input("Enter choice: "))
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))

    if choice == 1:
        print("Result:", a + b)
    elif choice == 2:
        print("Result:", a - b)
    elif choice == 3:
        print("Result:", a * b)
    elif choice == 4:
        try:
            print("Result:", a / b)
        except ZeroDivisionError:
            print("Error: Division by zero not allowed")
    else:
        print("Invalid choice")

calculator()

# 9) Create a file and store student details
def create_student_file():
    print("\n9) Creating student file")
    f = open("students.txt", "w")
    n = int(input("Enter number of students: "))
    for i in range(n):
        name = input("Enter student name: ")
        marks = input("Enter marks: ")
        f.write(name + " " + marks + "\n")
    f.close()
    print("File created successfully")
create_student_file()

# 10) Read data from file
def read_student_file():
    print("\n10) Reading student file")
    try:
        f = open("students.txt", "r")
        print(f.read())
        f.close()
    except FileNotFoundError:
        print("File not found")
read_student_file()

# 11) Exception handling - division by zero
def safe_division():
    print("\n11) Safe division")
    a = float(input("Enter numerator: "))
    b = float(input("Enter denominator: "))

    try:
        print("Result:", a / b)
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")
safe_division()

# 12) Student class
class Student:
    def __init__(self):
        self.name = input("Enter student name: ")

        try:
            self.marks = int(input("Enter marks: "))
        except:
            self.marks = 0
            print("Invalid marks, set to 0")

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)


s1 = Student()
s1.display()
```
