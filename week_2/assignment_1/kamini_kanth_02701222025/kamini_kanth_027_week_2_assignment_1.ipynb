```python

# NETFLIX USER ANALYTICS - COMPLETE MACHINE LEARNING PIPELINE

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score, confusion_matrix

# --- MOCK DATASET SETUP ---
# Creating a dummy dataset based on the description of 750 users [cite: 53, 54, 57]
np.random.seed(42)
n_users = 750

mock_data = {
    'UserID': [i for i in range(1001, 1001 + n_users)],
    'Age': np.random.randint(18, 65, size=n_users),
    'Gender': np.random.choice(['Male', 'Female'], size=n_users),
    'Subscription Type': np.random.choice(['Basic', 'Premium', 'VIP'], size=n_users),
    'WatchHours PerWeek': np.random.randint(5, 40, size=n_users),
    'Devices Used': np.random.randint(1, 6, size=n_users),
    'FavoriteGenre': np.random.choice(['Action', 'Comedy', 'Drama', 'Sci-Fi'], size=n_users),
    'AdClicks': np.random.randint(0, 15, size=n_users),
    'Monthly Spend': np.random.uniform(8.99, 25.99, size=n_users).round(2),
    'Subscription Renewed': np.random.choice(['Yes', 'No'], size=n_users, p=[0.7, 0.3])
}

df = pd.DataFrame(mock_data)


# ==============================================================================
# PART A: DATASET UNDERSTANDING [cite: 59]
# ==============================================================================

# Q1. Load the dataset and display the first five records. [cite: 60, 61]
print("--- Q1: First 5 Records ---")
print(df.head(), "\n")

# Q2. Determine the number of rows and columns in the dataset. [cite: 62, 63]
print("--- Q2: Dataset Shape ---")
rows, cols = df.shape
print(f"Rows: {rows}, Columns: {cols}\n")

# Q3. Display all column names. [cite: 64, 65]
print("--- Q3: Column Names ---")
print(list(df.columns), "\n")

# Q4. Identify numerical and categorical features. [cite: 66, 67]
print("--- Q4: Feature Types ---")
numerical_features = list(df.select_dtypes(include=[np.number]).columns)
categorical_features = list(df.select_dtypes(include=['object']).columns)
print(f"Numerical Features: {numerical_features}")
print(f"Categorical Features: {categorical_features}\n")

# Q5. Check whether the dataset contains missing values. [cite: 68, 69]
print("--- Q5: Missing Values Check ---")
print(df.isnull().sum(), "\n")


# ==============================================================================
# PART B: EXPLORATORY DATA ANALYSIS [cite: 70]
# ==============================================================================

# Q6. Calculate the average age of users. [cite: 71, 72]
print("--- Q6: Average Age ---")
print(f"{df['Age'].mean():.2f} years\n")

# Q7. Determine the average watch hours per week. [cite: 73, 74]
print("--- Q7: Average Watch Hours ---")
print(f"{df['WatchHours PerWeek'].mean():.2f} hours/week\n")

# Q8. Find the average monthly spending of users. [cite: 75, 76]
print("--- Q8: Average Monthly Spend ---")
print(f"${df['Monthly Spend'].mean():.2f}\n")

# Q9. Count the number of users in each subscription category. [cite: 77, 78]
print("--- Q9: Subscription Type Counts ---")
print(df['Subscription Type'].value_counts(), "\n")

# Q10. Determine the percentage of users who renewed their subscriptions. [cite: 79, 80]
print("--- Q10: Renewal Percentage ---")
renewal_pct = (df['Subscription Renewed'].value_counts(normalize=True) * 100)['Yes']
print(f"{renewal_pct:.2f}% of users renewed their subscription\n")


# ==============================================================================
# PART C: DATA PREPARATION [cite: 81]
# ==============================================================================

# Q11. Convert categorical features into numerical form. [cite: 82, 83]
# Using One-Hot Encoding for input features and mapping the target label
df_encoded = pd.get_dummies(df, columns=['Gender', 'Subscription Type', 'FavoriteGenre'], drop_first=True)
df_encoded['Subscription Renewed'] = df_encoded['Subscription Renewed'].map({'Yes': 1, 'No': 0})

print("--- Q11: Categorical Encoding Matrix Shape ---")
print(f"New DataFrame shape after encoding: {df_encoded.shape}\n")

# Q12. Define the feature set (X) and target variable (y) for subscription renewal prediction. [cite: 84, 85]
X_cls = df_encoded.drop(columns=['UserID', 'Subscription Renewed'])
y_cls = df_encoded['Subscription Renewed']

print("--- Q12: Feature and Target Dimensions ---")
print(f"X (Features) shape: {X_cls.shape} | y (Target) shape: {y_cls.shape}\n")

# Q13. Split the dataset into training and testing sets. [cite: 86, 87]
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_cls, y_cls, test_size=0.2, random_state=42)

print("--- Q13: Train/Test Split Sizes ---")
print(f"Training features: {X_train_c.shape}, Testing features: {X_test_c.shape}\n")


# ==============================================================================
# PART D: DECISION TREE CLASSIFICATION [cite: 88]
# ==============================================================================

# Q14. Train a Decision Tree model to predict whether a user will renew their subscription. [cite: 89, 90]
dt_model = DecisionTreeClassifier(max_depth=5, random_state=42)
dt_model.fit(X_train_c, y_train_c)
print("--- Q14: Decision Tree Training ---")
print("Decision Tree Classifier model successfully trained.\n")

# Q15. Evaluate the model using accuracy. [cite: 91, 92]
dt_preds = dt_model.predict(X_test_c)
dt_accuracy = accuracy_score(y_test_c, dt_preds)
print("--- Q15: Decision Tree Accuracy ---")
print(f"Accuracy Score: {dt_accuracy * 100:.2f}%\n")

# Q16. Generate and interpret the confusion matrix. [cite: 93, 94]
print("--- Q16: Confusion Matrix Output ---")
cm_dt = confusion_matrix(y_test_c, dt_preds)
print(cm_dt)
print("Interpretation Layout: [[True Negatives, False Positives], [False Negatives, True Positives]]\n")


# ==============================================================================
# PART E: K-NEAREST NEIGHBORS (KNN) [cite: 95]
# ==============================================================================

# Q17. Train a KNN classifier with K=5. [cite: 96, 97]
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train_c, y_train_c)
knn_preds = knn_model.predict(X_test_c)
print("--- Q17: KNN Training ---")
print("KNN Classifier model successfully trained.\n")

# Q18. Compare the accuracy of KNN with the Decision Tree model. [cite: 98, 99]
knn_accuracy = accuracy_score(y_test_c, knn_preds)
print("--- Q18: Model Performance Comparison ---")
print(f"Decision Tree Classification Accuracy: {dt_accuracy * 100:.2f}%")
print(f"K-Nearest Neighbors Classification Accuracy: {knn_accuracy * 100:.2f}%")
print(f"Winner: {'Decision Tree' if dt_accuracy > knn_accuracy else 'KNN'}\n")


# ==============================================================================
# PART F: LINEAR REGRESSION [cite: 100]
# ==============================================================================

# Q19. Train a Linear Regression model to predict monthly spending. [cite: 101, 102]
# For this task, Monthly Spend becomes our target target variable (y)
X_reg = df_encoded.drop(columns=['UserID', 'Monthly Spend'])
y_reg = df_encoded['Monthly Spend']

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)

reg_model = LinearRegression()
reg_model.fit(X_train_r, y_train_r)
print("--- Q19: Linear Regression Training ---")
print("Linear Regression model successfully fitted to dataset.\n")

# Q20. Predict the monthly spending for a new user and interpret the result. [cite: 103, 104]
print("--- Q20: Testing New User Prediction Output ---")
# Use the first row of test set data as a baseline example for evaluation
sample_user = X_test_r.iloc[[0]]
predicted_spend = reg_model.predict(sample_user)[0]
actual_spend = y_test_r.iloc[0]

print(f"Predicted Monthly Outlay Value: ${predicted_spend:.2f}")
print(f"Actual Value Stored in Data Matrix: ${actual_spend:.2f}")
print("Interpretation: The model tracks feature correlations to estimate a continuous financial trajectory value.\n")


# ==============================================================================
# BUSINESS REFLECTION QUESTIONS ANSWERS [cite: 105]
# ==============================================================================
"""
--- Business Reflection Insights Summary ---

1. Which factors appear to influence subscription renewal the most? [cite: 106]
   - Features tracking active system engagement like WatchHours PerWeek and metric trends
     spanning AdClicks tend to exhibit strong descriptive dependencies during classification splits.

2. Why is subscription renewal a classification problem? [cite: 107]
   - Because the final variable targets discrete outcome classes (Yes or No / 1 or 0)[cite: 57],
     rather than calculating data steps across infinite scale spectrum limits.

3. Why is monthly spending a regression problem? [cite: 108]
   - The tracked target space represents a continuous numerical balance scaling linearly[cite: 57],
     requiring error evaluation metrics based on exact real numbers instead of binary buckets.

4. Which algorithm performed better for renewal prediction? [cite: 109]
   - Check the printed accuracies in Q18. Often, Decision Trees perform better on unstructured
     categorical feature layouts due to logical split isolation behaviors.

5. How could the platform use these predictions to improve customer retention? [cite: 110]
   - The analytics framework can flag users returning low calculated renewal probabilities,
     triggering target distributions like custom promotional discounts or targeted genre alerts.
"""
print("==============================================================================")
print("Analytics Pipeline Executed and Ready for GitHub Code Review.")
print("==============================================================================")
```
