```python
# ==============================================================================
# WEEK 2 ASSIGNMENT 2 - SUBMISSION
# ==============================================================================

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import confusion_matrix
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier


# ==============================================================================
# Q1. Loan Application Prediction Strategy
# ==============================================================================
# Scenario: A bank wants to predict if a loan is approved or rejected.
# Type: This is a CLASSIFICATION problem because the target outcome is categorical (Yes/No).
# Algorithm: Decision Trees or KNN are suitable. Linear Regression is incorrect here.

# Demonstration using a mock loan evaluation dataset:
loan_data = {
    'Monthly_Income': [4500, 8000, 2500, 11000, 5200, 3100],
    'Credit_Score': [610, 780, 520, 810, 690, 580],
    'Loan_Approved': [0, 1, 0, 1, 1, 0]  # 0 = Rejected, 1 = Approved
}
df_loans = pd.DataFrame(loan_data)

# Splitting features and target label
X_loans = df_loans[['Monthly_Income', 'Credit_Score']]
y_loans = df_loans['Loan_Approved']

# Training a classification model as the solution
loan_model = DecisionTreeClassifier(random_state=42)
loan_model.fit(X_loans, y_loans)

print("--- Q1 Output ---")
print(f"Model successfully trained for binary classification.")
print(f"Features shapes: {X_loans.shape}, Label shape: {y_loans.shape}\n")


# ==============================================================================
# Q2. Feature vs Label Explanation
# ==============================================================================
# Feature: The independent input variables used to train the model.
# Label: The dependent target variable that the model is trying to predict.
# Real-World Example: Weather App tracking conditions to predict rain.

weather_data = {
    'Temperature_C': [22, 31, 15, 28],      # Feature 1
    'Humidity_Pct': [85, 40, 90, 45],       # Feature 2
    'Wind_Speed_Kmh': [12, 5, 25, 8],       # Feature 3
    'Will_Rain': ['Yes', 'No', 'Yes', 'No'] # Target Label
}
df_weather = pd.DataFrame(weather_data)

print("--- Q2 Output ---")
print("Features (Inputs):\n", df_weather[['Temperature_C', 'Humidity_Pct', 'Wind_Speed_Kmh']])
print("\nLabel (Target Output):\n", df_weather['Will_Rain'])
print("\n")


# ==============================================================================
# Q3. Decision Tree Traceability
# ==============================================================================
# A Decision Tree reaches predictions by breaking down a dataset into smaller
# subsets based on feature values, forming tree branches and leaf nodes.

# Re-using loan data to display the logical execution path:
dt_rules = DecisionTreeClassifier(max_depth=2, random_state=42)
dt_rules.fit(X_loans, y_loans)

print("--- Q3 Output ---")
print("Execution structure generated by the Decision Tree:")
print(export_text(dt_rules, feature_names=['Monthly_Income', 'Credit_Score']))


# ==============================================================================
# Q4 & Q5. KNN Logic and K=5 Voting Execution
# ==============================================================================
# KNN is supervised because it maps input coordinates to known, historical labels.
# If labels are missing, it cannot classify (you would use K-Means instead).
# Given K=5 neighbors: ['Yes', 'Yes', 'No', 'Yes', 'No'], the system evaluates
# the mode (majority vote). 3 'Yes' vs 2 'No' results in a 'Yes' prediction.

# Simulating the exact neighborhood vote calculation:
neighbors_votes = np.array(['Yes', 'Yes', 'No', 'Yes', 'No'])

# Count occurrences programmatically
unique_outputs, vote_counts = np.unique(neighbors_votes, return_counts=True)
winning_index = np.argmax(vote_counts)
final_prediction = unique_outputs[winning_index]

print("--- Q4 & Q5 Output ---")
print(f"Votes registered: {neighbors_votes}")
print(f"Calculated evaluation outcome: {final_prediction} (Majority vote win)")
print("\n")


# ==============================================================================
# Q6. Linear Regression Equation Analysis
# ==============================================================================
# Equation: y = mx + b
# y: Predicted dependent value
# x: Independent input feature
# m: Slope (weight coefficient showing scaling factor)
# b: Y-intercept (bias value when input x is 0)

# Dummy linear relationship dataset creation (y = 2x + 1)
X_linear = np.array([[1], [2], [3], [4]])
y_linear = np.array([3, 5, 7, 9])

lin_reg = LinearRegression()
lin_reg.fit(X_linear, y_linear)

print("--- Q6 Output ---")
print(f"Extracted Formula: y = ({lin_reg.coef_[0]:.1f})x + {lin_reg.intercept_:.1f}")
print("\n")


# ==============================================================================
# Q7. Correlation Space Implementation
# ==============================================================================
# Correlation defines the linear relationship strength and direction between vectors.
#  +1 -> Perfect positive linear movement
#   0 -> No linear correlation detected
#  -1 -> Perfect negative linear movement

correlation_matrix_data = {
    'Feature_Base': [1, 2, 3, 4, 5],
    'Pos_1': [2, 4, 6, 8, 10],   # Perfect positive movement
    'Neg_Minus1': [5, 4, 3, 2, 1], # Perfect negative movement
    'Zero_0': [1, 5, 2, 4, 3]      # Chaotic / No clear linear correlation
}
df_matrix = pd.DataFrame(correlation_matrix_data)

print("--- Q7 Output ---")
print("Computed Correlation Matrix:")
print(df_matrix.corr())
print("\n")


# ==============================================================================
# Q8 & Q9. Disease Evaluation Matrix vs Correlation Matrix
# ==============================================================================
# Scenario: Actual Condition = Disease Present, Prediction = Healthy.
# Type: False Negative (Type II Error). This is dangerous because an ill patient
# left untreated faces risk of medical degradation or mortality.
# Confusion Matrix focuses on cross-tabulating factual vs predicted classification tallies.
# Correlation Matrix checks variable feature-space linear dependencies.

# Mapping: 1 = Disease Present, 0 = Healthy
y_true_disease = [1, 1, 0, 0, 1]
y_pred_disease = [0, 1, 0, 0, 1]  # Index 0 holds the tracked False Negative instance

conf_matrix = confusion_matrix(y_true_disease, y_pred_disease)
tn, fp, fn, tp = conf_matrix.ravel()

print("--- Q8 & Q9 Output ---")
print("Evaluated Confusion Matrix Arrays:")
print(conf_matrix)
print(f"Quantified False Negatives (Dangerous Misses): {fn}")
print("\n")


# ==============================================================================
# Q10. Machine Learning Algorithm Benchmarking Pipeline
# ==============================================================================
# Summary definitions of standard architectural concepts:
# - Decision Tree: Splits feature space iteratively based on optimization rules.
# - KNN: Maps classifications based on feature proximity data configurations.
# - SVM: Searches for maximum boundary hyperplanes separating target classes.
# - Linear Regression: Finds continuous trends by optimizing linear fit metrics.
# - Neural Network: Passes input through hidden nodes to identify complex patterns.

# Running multiple algorithms through a comparative interface:
X_synthetic = np.random.rand(40, 2)
y_synthetic = np.random.randint(0, 2, size=40)

model_dictionary = {
    "Decision Tree Classifier": DecisionTreeClassifier(random_state=42),
    "K-Nearest Neighbors": KNeighborsClassifier(n_neighbors=3),
    "Support Vector Machine": SVC(kernel='linear'),
    "Linear Proxy (Logistic)": LogisticRegression(),
    "Multi-Layer Perceptron NN": MLPClassifier(max_iter=300, random_state=42)
}

print("--- Q10 Output ---")
for model_name, instantiated_model in model_dictionary.items():
    instantiated_model.fit(X_synthetic, y_synthetic)
    model_score = instantiated_model.score(X_synthetic, y_synthetic)
    print(f"[{model_name}] Training Initialization Check Accuracy: {model_score*100:.1f}%")

print("\n==============================================================================")
print("Processing Completed Successfully.")
print("==============================================================================")
```
