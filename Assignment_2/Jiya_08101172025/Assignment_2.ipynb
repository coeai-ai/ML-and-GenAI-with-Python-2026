```python
n = int(input("Enter a number: "))

fact = 1

for i in range(1, n + 1):
    fact *= i

print("Factorial =", fact)
```

```python
name = input("Enter Name: ")

m1 = float(input("Maths: "))
m2 = float(input("Physics: "))
m3 = float(input("Chemistry: "))

percentage = (m1 + m2 + m3) / 3

if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
else:
    grade = "Fail"

print("\nResult")
print("Name:", name)
print("Percentage:", percentage)
print("Grade:", grade)
```
