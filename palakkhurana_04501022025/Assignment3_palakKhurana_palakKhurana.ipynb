```python
#sum of 1st 10 natural numbers
def sum(n):
    sum = 0
    for i in range(1,n+1):
        sum = sum + i
    print (sum)
sum(8)
```

```python
#reversing a number
def reverse(num):
    reversed_num = 0

    while num > 0:
        digit = num % 10                
        reversed_num = (reversed_num * 10) + digit  
        num = num // 10                   

    print(reversed_num)  # Output: 54321
reverse(12345)
```

```python
#count digits in a number
def count_digit(num):
    count = 0
    while num > 0:
        num = num // 10
        count = count + 1
    print(count)
count_digit(4636)
```

```python
#check pallindrome
def check_pallindrome(num):
    reversed_num = 0
    original_num = num
    while num > 0:
        digit = num % 10                
        reversed_num = (reversed_num * 10) + digit  
        num = num // 10                   
    if(reversed_num==original_num):
        print("pallindrome") 
    else:
        print("not a pallindrome")
check_pallindrome(45650)
```

```python
#calculator
def calculator(operation,a,b):
    if(operation == '+'):
        print(a+b)
    if(operation == '-'):
        print(a-b)
    if(operation == '*'):
        print(a*b)
    if(operation == '/'):
        print(a/b)
calculator('+',34,42)
```

```python
# Create a Student class with name and marks
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f"Name: {self.name}")
        print(f"Marks: {self.marks}")
s1 = Student("Palak", 85)
s2 = Student("Riya", 92)

s1.display()
print()
s2.display()
```
