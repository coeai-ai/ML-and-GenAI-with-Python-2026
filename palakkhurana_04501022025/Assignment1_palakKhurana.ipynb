```python

# --- Step 1: Collect Student Information ---
student_name = input("Enter student's name: ")
roll_number = input("Enter roll number: ")
class_name = input("Enter class/grade: ")

# --- Step 2: Input Marks for 5 Subjects ---
# Marks are converted to floats to allow decimal values
math_marks = float(input("Enter Math marks (out of 100): "))
science_marks = float(input("Enter Science marks (out of 100): "))
english_marks = float(input("Enter English marks (out of 100): "))
history_marks = float(input("Enter History marks (out of 100): "))
art_marks = float(input("Enter Art marks (out of 100): "))

# --- Step 3: Perform Calculations ---
# Calculate the sum of all subject marks
total_marks = math_marks + science_marks + english_marks + history_marks + art_marks

# Calculate percentage (assuming each subject is out of 100, max marks = 500)
max_marks = 500
percentage = (total_marks / max_marks) * 100
print("total_marks of student",total_marks)
print("percentage",percentage)
```
